```python
import pandas as pd
import numpy as np

# 1. Load your actual cafe sales data
df_cafe = pd.read_csv('/content/automated-ecommerce-pipeline/dirty-cafe-sales-csv (2).csv')

print("--- Original Data Sample ---")
print(df_cafe[['Transaction ID', 'Item', 'Total Spent', 'Transaction Date']].head())

# 2. Run standard cleaning steps on it
# Replace text errors with proper NaN values
df_cafe.replace(['ERROR', 'UNKNOWN', 'NaN'], np.nan, inplace=True)

# Clean up numeric columns
df_cafe['Quantity'] = pd.to_numeric(df_cafe['Quantity'], errors='coerce')
df_cafe['Price Per Unit'] = pd.to_numeric(df_cafe['Price Per Unit'], errors='coerce')

# Recalculate 'Total Spent' where it was missing or errored
df_cafe['Total Spent'] = df_cafe['Quantity'] * df_cafe['Price Per Unit']

# Clean up dates
df_cafe['Transaction Date'] = pd.to_datetime(df_cafe['Transaction Date'], errors='coerce')

print("\n--- Cleaned Data Sample ---")
print(df_cafe.head())

# 3. Save your true cleaned data to a new CSV file
df_cafe.to_csv('/content/cleaned_cafe_sales.csv', index=False)
print("\nSuccess! Your actual cleaned dataset is saved to 'cleaned_cafe_sales.csv'")
```

```python
%%writefile ecommerce_etl_pipeline.py
import pandas as pd
import numpy as np
import sqlite3
import argparse

def run_pipeline(input_file):
    print("Extracting...")
    # Read the provided CSV file
    df = pd.read_csv(input_file)

    print("Transforming...")
    # Clean up standard data errors
    df.replace(['ERROR', 'UNKNOWN', 'NaN'], np.nan, inplace=True)

    # Cast and fix metrics
    df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
    df['Price Per Unit'] = pd.to_numeric(df['Price Per Unit'], errors='coerce')
    df['Total Spent'] = df['Quantity'] * df['Price Per Unit']
    df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], errors='coerce')

    print("Validating and Loading...")
    # Connect to the local database and save the actual cleaned records
    conn = sqlite3.connect('ecommerce_data.db')
    df.to_sql('cafe_sales', conn, if_exists='replace', index=False)
    conn.close()

    print(f"Success: Data validated and loaded into table 'cafe_sales'.")
    print(f"Pipeline finished. Total records: {len(df)}")

if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument('--input', type=str, required=True)
    args = parser.parse_args()
    run_pipeline(args.input)
```
